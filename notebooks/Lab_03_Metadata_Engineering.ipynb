```python
from pathlib import Path
import pandas as pd
```

```python
# Project root (where the GitHub repository is)
PROJECT_ROOT = Path.cwd()

# Dataset paths
DATASET_PATH = PROJECT_ROOT / "data" / "raw" / "FaceForensics++_C23"
CSV_FOLDER = DATASET_PATH / "csv"

# Output folder
OUTPUT_FOLDER = PROJECT_ROOT / "data" / "metadata"
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)
```

```python
OUTPUT_FOLDER.mkdir(parents=True, exist_ok=True)
```

```python
CSV_FILES = [
    "original.csv",
    "Deepfakes.csv",
    "FaceSwap.csv",
    "Face2Face.csv",
    "FaceShifter.csv",
    "NeuralTextures.csv",
    "DeepFakeDetection.csv"
]
```

```python
def load_csv_files():
    pass
```

```python
def load_csv_files():
    """
    Read all required CSV files
    and combine them into one DataFrame.
    """

    dataframes = []

    for file_name in CSV_FILES:
        file_path = CSV_FOLDER / file_name

        print(f"Reading {file_name}...")

        df = pd.read_csv(file_path)

        dataframes.append(df)

    master_df = pd.concat(dataframes, ignore_index=True)

    return master_df
```

```python
def clean_metadata(df):
    """
    Clean the metadata by removing
    unnecessary columns.
    """

    columns_to_drop = [
        "Unnamed: 0",
        "Codec",
        "File Size(MB)"
    ]

    df = df.drop(columns=columns_to_drop, errors="ignore")

    return df
```

```python
metadata = load_csv_files()

cleaned_metadata = clean_metadata(metadata)

print(cleaned_metadata.head())

print(cleaned_metadata.shape)
```

```python
def save_metadata(df):
    """
    Save the cleaned metadata
    as master_metadata.csv
    """

    output_file = OUTPUT_FOLDER / "master_metadata.csv"

    df.to_csv(output_file, index=False)

    print(f"Metadata saved successfully to:\n{output_file}")
```

```python
metadata = load_csv_files()

cleaned_metadata = clean_metadata(metadata)

save_metadata(cleaned_metadata)
```
